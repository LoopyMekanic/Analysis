```python
# Import Pandas and Numpy
import pandas as pd
import numpy as np
```

```python
# Create Pandas Series from list (column in table)
my_list = [5,9,12,27]
# Create a series
my_series = pd.Series(my_list)
```

```python
# Print without formatting
print(my_series)
```

```python
# Print with formatting
my_series
```

```python
# Pull item from series with index number
print(my_series[1])
my_series[1]
```

```python
# Create labels with index arguementny__index = ["A", "B", "C", "D"]
my_index = ["A", "B", "C", "D"]
my_series2 = pd.Series(my_list, my_index)
my_series2
```

```python
# Create labels manually
my_series3 = pd.Series(my_list, ["E", "F", "G", "H"])
my_series3
```

```python
# Pull item from series with label
# print(my_series3.iloc[3])
print(my_series3["G"])
```
